### HBN CPAC (CBIC / RU / CUNY only — SI excluded; run-1 → run-2 fallback; NO_ACQ)

```python
overlap_hbn = set(split_dict["hbn_age_trainval_subjects"]) | set(split_dict["hbn_age_test_subjects"])
print(f"HBN subjects to process: {len(overlap_hbn)}")
```

```python
# HBN: 3T sites only (CBIC / RU / CUNY). SI excluded (1.5T Avanto, incompatible TR).
# Each subject lives at exactly one site; we try them all but only one will resolve.
# Run-1 is preferred; fall back to run-2 only if run-1 isn't available for this subject.
# NO_ACQ: exact paths (no wildcards) only match the non-variant filename.
dataset = dl.Dataset("HBN_CPAC")

HBN_SITES = ["CBIC", "RU", "CUNY"]
SCALES = [
    "Schaefer2018p200n17",
    "Schaefer2018p300n17",
    # "Schaefer2018p400n17",
    # "Schaefer2018p1000n17",
]

def hbn_path(sub, site, run, scale):
    return rf"{dir_path}\HBN_CPAC\cpac_RBCv0\sub-{sub}\ses-HBNsite{site}\func\sub-{sub}_ses-HBNsite{site}_task-rest_{run}_atlas-{scale}_space-MNI152NLin6ASym_reg-36Parameter_desc-PartialNilearn_correlations.tsv"

failures = []
downloaded = []
for sub in tqdm(overlap_hbn):
    all_found = True
    for scale in SCALES:
        got_this_scale = False
        # Try run-1 across sites first; only fall back to run-2 if run-1 yields nothing.
        for run in ["run-1", "run-2"]:
            if got_this_scale:
                break
            for site in HBN_SITES:
                path = hbn_path(sub, site, run, scale)
                try:
                    dataset.get(path, on_failure="ignore")
                except Exception:
                    pass
                if os.path.exists(path) and os.path.getsize(path) > 1000:
                    got_this_scale = True
                    break  # single file per (subject, scale)
        if not got_this_scale:
            all_found = False
    (downloaded if all_found else failures).append(sub)

print(f"\nDownloaded: {len(downloaded)}, Failed: {len(failures)}")
```
